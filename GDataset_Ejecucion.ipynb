<h2 style="margin: 0; color: #818cf8; font-size: 1.35rem; font-weight: 700;">Fase 2: Curación de Datos y Reglas de Negocio (Data Curation)</h2>
</div>
<p style="margin: 0 0 16px 0; line-height: 1.6; color: #cbd5e1; font-size: 0.95rem;">
Bienvenidos a la realidad de los datasets crudos de visión por computador: <strong>"Garbage In, Garbage Out"</strong>. Los repositorios públicos abiertos suelen incluir etiquetas superpuestas, clases ambiguas, inconsistencias taxonómicas y ruido de fondo. Si alimentamos esto directamente a una red convolucional, el modelo colapsará en falsos positivos. Implementamos un <strong>Pipeline Unificado de Limpieza</strong> en memoria con tres reglas estrictas:
</p>
<div style="display: grid; grid-template-columns: repeat(auto-fit, minmax(260px, 1fr)); gap: 12px; margin-bottom: 16px;">
<div style="background: rgba(15, 23, 42, 0.6); border: 1px solid rgba(129, 140, 248, 0.2); border-radius: 8px; padding: 14px 16px;">
<strong style="color: #a5b4fc; font-size: 0.9rem;">1. Unificación Semántica</strong>
<p style="margin: 6px 0 0 0; color: #94a3b8; font-size: 0.84rem; line-height: 1.5;">Fusione de clases equivalentes. Por ejemplo, <code>Bus</code> (ID 48) es estructuralmente homogéneo a <code>Car</code> (ID 6) para nuestro objetivo de movilidad. Reducimos la dimensionalidad redundante.</p>
</div>
<div style="background: rgba(15, 23, 42, 0.6); border: 1px solid rgba(129, 140, 248, 0.2); border-radius: 8px; padding: 14px 16px;">
<strong style="color: #a5b4fc; font-size: 0.9rem;">2. Purga de Outliers</strong>
<p style="margin: 6px 0 0 0; color: #94a3b8; font-size: 0.84rem; line-height: 1.5;">Eliminación de anotaciones ultrasubjetivas o parciales (ej. <code>Human face</code> - ID 15) que solo introducen colisiones de gradiente con la clase principal <code>Person</code>.</p>
</div>
<div style="background: rgba(15, 23, 42, 0.6); border: 1px solid rgba(129, 140, 248, 0.2); border-radius: 8px; padding: 14px 16px;">
<strong style="color: #a5b4fc; font-size: 0.9rem;">3. Poda de Background Noise</strong>
<p style="margin: 6px 0 0 0; color: #94a3b8; font-size: 0.84rem; line-height: 1.5;">Descarte agresivo de imágenes vacías o compuestas únicamente por elementos irrelevantes de fondo, evitando que la red aprenda a predecir ruido de clase nula.</p>
</div>
</div>
</div>


```python
import os
import yaml
import shutil

data_path = "GData"
dataset_yaml_path = f"{data_path}/dataset.yaml"

target_classes_names = [
    "Human body", "Person", "Dog", "Cat", "Bicycle", 
    "Car", "Bus", "Motorcycle", "Bird", "Building", "Tree", "Human face"
]

# 1. Mapeo de IDs
with open(dataset_yaml_path, "r") as f:
    dataset_yaml = yaml.safe_load(f)

name_to_id = {name: id for id, name in dataset_yaml["names"].items()}
target_class_ids = {name_to_id[name] for name in target_classes_names if name in name_to_id}

# IDs Dinámicos por nombre
person_id = name_to_id.get("Person", -1)
human_body_id = name_to_id.get("Human body", -1)
bus_id = name_to_id.get("Bus", -1)
car_id = name_to_id.get("Car", -1)
human_face_id = name_to_id.get("Human face", -1)
tree_id = name_to_id.get("Tree", -1)

print(f"Total de IDs válidos a conservar: {len(target_class_ids)}")

# 2. Configurar carpetas (Borrar clean_labels si existe para empezar de cero)
labels_dir = os.path.join(data_path, "labels")
clean_labels_dir = os.path.join(data_path, "clean_labels")

if os.path.exists(clean_labels_dir):
    print("Borrando la carpeta 'clean_labels' existente para recrearla...")
    shutil.rmtree(clean_labels_dir)
os.makedirs(clean_labels_dir)

# Contadores
procesados = 0
guardados = 0
descartados = 0
anotaciones_guardadas = 0

# 3. Pipeline Unificado
for split in os.listdir(labels_dir):
    split_dir = os.path.join(labels_dir, split)
    if not os.path.isdir(split_dir): continue
        
    clean_split_dir = os.path.join(clean_labels_dir, split)
    os.makedirs(clean_split_dir, exist_ok=True)
        
    for label_file in os.listdir(split_dir):
        if not label_file.endswith(".txt"): continue
            
        original_path = os.path.join(split_dir, label_file)
        new_path = os.path.join(clean_split_dir, label_file)
        
        with open(original_path, "r") as f:
            lines = f.readlines()
            
        procesados += 1
        
        # Primero procesamos las líneas para saber qué clases estarán en el archivo final
        clean_lines = []
        classes_present = set()
        
        for line in lines:
            parts = line.strip().split()
            if not parts: continue
            class_id = int(parts[0])
            
            # Filtro 1: Descartar todo lo que no esté en nuestra lista de target
            if class_id not in target_class_ids:
                continue
                
            # Filtro 2: Convertir Human body a Person
            if class_id == human_body_id and person_id != -1:
                parts[0] = str(person_id)
                class_id = person_id
                line = " ".join(parts) + "\n"
                
            # Filtro 3: Convertir Bus a Car
            if class_id == bus_id and car_id != -1:
                parts[0] = str(car_id)
                class_id = car_id
                line = " ".join(parts) + "\n"
                
            # Filtro 4: Ignorar las líneas de la clase Human face
            if class_id == human_face_id:
                classes_present.add(human_face_id)
                continue # No la añadimos a clean_lines
                
            classes_present.add(class_id)
            clean_lines.append(line)
            
        # Descartar si el archivo final quedó vacío
        if not clean_lines:
            descartados += 1
            continue
            
        # Si estuvo Human face, pero no estuvo Person (Human body ya se convirtió a Person arriba)
        # Entonces se descarta todo el archivo.
        if human_face_id in classes_present and person_id not in classes_present:
            descartados += 1
            continue
            
        # Actualizamos las clases presentes post-limpieza
        final_classes = set([int(l.strip().split()[0]) for l in clean_lines])
        
        # Descartar si ÚNICAMENTE tiene la clase Person
        if len(final_classes) == 1 and person_id in final_classes:
            descartados += 1
            continue
            
        # Descartar si ÚNICAMENTE tiene la clase Tree
        if len(final_classes) == 1 and tree_id in final_classes:
            descartados += 1
            continue
            
        # Descartar si ÚNICAMENTE tiene la clase Car
        if len(final_classes) == 1 and car_id in final_classes:
            descartados += 1
            continue
            
        # Si pasó todos los filtros, guardamos el archivo
        with open(new_path, "w") as f:
            f.writelines(clean_lines)
            
        guardados += 1
        anotaciones_guardadas += len(clean_lines)

print("\n✅ Pipeline Unificado de Limpieza Completado")
print(f"Archivos originales procesados: {procesados}")
print(f"Archivos guardados en clean_labels: {guardados}")
print(f"Archivos completamente descartados por reglas de limpieza: {descartados}")
print(f"Anotaciones finales conservadas: {anotaciones_guardadas}")
```

<h2 style="margin: 0; color: #34d399; font-size: 1.35rem; font-weight: 700;">Fase 3: Exploratory Data Analysis (EDA)</h2>
</div>
<p style="margin: 0 0 14px 0; line-height: 1.6; color: #cbd5e1; font-size: 0.95rem;">
En Computer Vision, entrenar a ciegas es un error metodológico crítico. Necesitamos auditar empíricamente la <strong>distribución de frecuencias y tamaños de cajas delimitadoras</strong> tras la depuración.
</p>
<div style="background: rgba(15, 23, 42, 0.6); border: 1px solid rgba(52, 211, 153, 0.2); border-radius: 8px; padding: 14px 18px; color: #94a3b8; font-size: 0.88rem; line-height: 1.6;">
<strong style="color: #6ee7b7;">¿Por qué es letal el Desbalance de Clases (*Class Imbalance*)?</strong><br>
El desbalance severo corrompe la dinámica de optimización durante el <em>backpropagation</em>. Si el dataset contiene <strong>40,000 instancias de <code>Person</code></strong> y apenas <strong>2,500 de <code>Cat</code></strong>, el gradiente minimizará la función de pérdida general ignorando la clase minoritaria: el modelo se convertirá en un detector perfecto de personas pero será completamente ciego ante gatos.
</div>

```python
import os
import yaml
from collections import Counter
import pandas as pd
from IPython.display import display

data_path = "GData"
dataset_yaml_path = f"{data_path}/dataset.yaml"

with open(dataset_yaml_path, "r") as f:
    dataset_yaml = yaml.safe_load(f)

class_names = dataset_yaml["names"]
clean_labels_dir = os.path.join(data_path, "clean_labels")

counter = Counter()

# Contar apariciones post-limpieza
for split in os.listdir(clean_labels_dir):
    split_dir = os.path.join(clean_labels_dir, split)
    if not os.path.isdir(split_dir): continue
        
    for label_file in os.listdir(split_dir):
        if not label_file.endswith(".txt"): continue
            
        with open(os.path.join(split_dir, label_file), "r") as f:
            for line in f:
                parts = line.strip().split()
                if parts:
                    counter[int(parts[0])] += 1

data = []
for class_id, count in counter.most_common():
    name = class_names.get(class_id, "Desconocido")
    data.append({"ID": class_id, "Clase": name, "Conteo": count})

print("Estadísticas Finales de Etiquetas Limpias:")
df_stats = pd.DataFrame(data)
display(df_stats)
```

```python
# ==============================================================================
# 🚩 CONFIGURACIÓN DE SANITY CHECK (VISUALIZACIÓN)
# ==============================================================================
# Cambia esta bandera a True para inspeccionar visualmente imágenes con sus bounding boxes.
# Por defecto se mantiene en False para permitir ejecuciones rápidas y headless.
ENABLE_SANITY_CHECK_VISUALIZATION = True
NUM_SAMPLES_TO_VISUALIZE = 4  # Cantidad de imágenes a desplegar

# ==============================================================================
# EJECUCIÓN DEL SANITY CHECK
# ==============================================================================
if not ENABLE_SANITY_CHECK_VISUALIZATION:
    print("⏸️  Sanity Check visual OMITIDO (ENABLE_SANITY_CHECK_VISUALIZATION = False).")
    print("💡 Para inspeccionar visualmente imágenes con sus bounding boxes dibujadas,")
    print("   cambia 'ENABLE_SANITY_CHECK_VISUALIZATION = True' arriba y vuelve a ejecutar esta celda.")
else:
    import os
    import random
    import yaml
    import matplotlib.pyplot as plt
    import matplotlib.patches as patches
    from PIL import Image

    data_path = "GData"
    dataset_yaml_path = f"{data_path}/dataset.yaml"
    clean_labels_dir = os.path.join(data_path, "clean_labels")

    # 1. Cargar catálogo de nombres de clases
    class_names = {}
    if os.path.exists(dataset_yaml_path):
        with open(dataset_yaml_path, "r") as f:
            yaml_content = yaml.safe_load(f)
            class_names = yaml_content.get("names", {})

    # Paleta de colores distintiva para las cajas delimitadoras
    palette = [
        "#00FF66", "#38BDF8", "#F43F5E", "#FBBF24", "#A855F7",
        "#FB923C", "#EC4899", "#10B981", "#3B82F6", "#F97316"
    ]

    # 2. Recolectar pares de (imagen, etiqueta) en clean_labels
    available_pairs = []
    if os.path.exists(clean_labels_dir):
        for split in os.listdir(clean_labels_dir):
            split_labels = os.path.join(clean_labels_dir, split)
            split_images = os.path.join(data_path, "images", split)
            if not os.path.isdir(split_labels):
                continue
            for f in os.listdir(split_labels):
                if f.endswith(".txt"):
                    base = os.path.splitext(f)[0]
                    for ext in [".jpg", ".jpeg", ".png"]:
                        img_candidate = os.path.join(split_images, base + ext)
                        if os.path.exists(img_candidate):
                            available_pairs.append((img_candidate, os.path.join(split_labels, f)))
                            break

    if not available_pairs:
        print("⚠️ No se encontraron pares de imágenes y etiquetas en GData/clean_labels.")
    else:
        # 3. Muestrear aleatoriamente
        random.seed(42)  # Semilla reproducible para consistencia
        samples = random.sample(available_pairs, min(NUM_SAMPLES_TO_VISUALIZE, len(available_pairs)))
        n_imgs = len(samples)
        cols = 2 if n_imgs > 1 else 1
        rows = (n_imgs + cols - 1) // cols

        fig, axes = plt.subplots(rows, cols, figsize=(7 * cols, 6 * rows), squeeze=False)
        plt.subplots_adjust(wspace=0.15, hspace=0.25)

        for idx, (img_path, label_path) in enumerate(samples):
            ax = axes[idx // cols][idx % cols]
            img = Image.open(img_path).convert("RGB")
            img_w, img_h = img.size
            ax.imshow(img)

            # Leer anotaciones YOLO (class_id x_c y_c w h)
            with open(label_path, "r") as f:
                lines = f.readlines()

            for line in lines:
                parts = line.strip().split()
                if len(parts) >= 5:
                    cls_id = int(parts[0])
                    xc, yc, w, h = map(float, parts[1:5])

                    # Conversión de coordenadas normalizadas YOLO a píxeles absolutos
                    box_w = w * img_w
                    box_h = h * img_h
                    x_min = (xc * img_w) - (box_w / 2)
                    y_min = (yc * img_h) - (box_h / 2)

                    color = palette[cls_id % len(palette)]
                    cls_name = class_names.get(cls_id, f"ID {cls_id}")

                    # Rectángulo delimitador
                    rect = patches.Rectangle(
                        (x_min, y_min), box_w, box_h,
                        linewidth=2.5, edgecolor=color, facecolor="none"
                    )
                    ax.add_patch(rect)

                    # Etiqueta badge de la clase
                    ax.text(
                        x_min, max(0, y_min - 6),
                        f" {cls_name} ",
                        fontsize=9, weight="bold", color="black",
                        bbox=dict(facecolor=color, edgecolor="none", pad=2, alpha=0.9)
                    )

            base_name = os.path.basename(img_path)
            ax.set_title(f"Muestra: {base_name} ({len(lines)} objetos)", fontsize=11, weight="semibold", pad=8)
            ax.axis("off")

        # Ocultar subplots vacíos si n_imgs es impar
        for idx in range(n_imgs, rows * cols):
            axes[idx // cols][idx % cols].axis("off")

        plt.show()
        print(f"✅ Sanity Check completado: {n_imgs} muestras visualizadas.")
```

<h2 style="margin: 0; color: #c084fc; font-size: 1.35rem; font-weight: 700;">Fase 5: Partición Estratificada y Formato YOLO (Stratified Split)</h2>
</div>
<p style="margin: 0 0 16px 0; line-height: 1.6; color: #cbd5e1; font-size: 0.95rem;">
En datasets <strong>multi-etiqueta fuertemente desbalanceados</strong>, una división aleatoria (<em>Random Split</em> 90/10) es un peligro: las clases raras podrían quedar confinadas 100% en <code>train</code> sin ninguna representación en <code>val</code>, cegando la evaluación del modelo.
</p>
<div style="background: rgba(15, 23, 42, 0.6); border: 1px solid rgba(192, 132, 252, 0.2); border-radius: 8px; padding: 14px 18px; margin-bottom: 16px; color: #cbd5e1; font-size: 0.9rem; line-height: 1.6;">
<strong style="color: #d8b4fe;">Algoritmo de Prioridad por Minoría (Minority Priority Split):</strong>
<ol style="margin: 8px 0 0 0; padding-left: 20px; color: #94a3b8; font-size: 0.85rem; line-height: 1.6;">
<li><strong>Ordenamiento por rareza:</strong> Clasificamos las clases globalmente de menor a mayor frecuencia empírica.</li>
<li><strong>Bucketización por clase más rara:</strong> Si una imagen tiene 20 personas (frecuentes) y 1 gato (raro), se cataloga en el bucket de "Gatos".</li>
<li><strong>Partición Aislada:</strong> Barajamos y cortamos la proporción 90/10 dentro de cada bucket por separado.</li>
</ol>
</div>
<div style="background: rgba(30, 41, 59, 0.4); border-left: 3px solid #38bdf8; padding: 10px 14px; border-radius: 6px; font-size: 0.84rem; color: #94a3b8; line-height: 1.5;">
<strong>Remapeo Contiguo de IDs:</strong> Convertimos las clases supervivientes a un vector continuo <code>[0, 1, 2, ..., N-1]</code>. YOLO requiere identificadores contiguos; índices discontinuos causarían un colapso catastrófico (<code>IndexError</code>) al calcular el tensor de pérdida.
</div>
</div>


```python
import os
import shutil
import random
import yaml
from collections import Counter
import pandas as pd
from IPython.display import display

data_path = "GData"
clean_labels_dir = os.path.join(data_path, "clean_labels")
images_dir = os.path.join(data_path, "images")
yolo_dataset_dir = os.path.join(data_path, "YOLO_Dataset")

# Fijar semilla aleatoria para reproducibilidad
random.seed(42)

print("Paso 1: Analizando frecuencias y asignando imágenes a sus clases minoritarias...")

# 1. Contar frecuencias globales y recopilar datos
global_counts = Counter()
file_data = {}

for split in os.listdir(clean_labels_dir):
    split_dir = os.path.join(clean_labels_dir, split)
    if not os.path.isdir(split_dir): continue
    for label_file in os.listdir(split_dir):
        if not label_file.endswith(".txt"): continue
        file_path = os.path.join(split, label_file) # ruta relativa
        classes_in_file = set()
        
        with open(os.path.join(clean_labels_dir, file_path), "r") as f:
            for line in f:
                parts = line.strip().split()
                if parts:
                    class_id = int(parts[0])
                    classes_in_file.add(class_id)
                    global_counts[class_id] += 1
                    
        if classes_in_file:
            file_data[file_path] = list(classes_in_file)

# Ordenar IDs de clases por frecuencia (ascendente, el más raro primero)
sorted_classes = [c for c, count in global_counts.most_common()[::-1]]

# Mapeo: ID viejo -> ID secuencial nuevo (0 a N)
old_ids = sorted(list(global_counts.keys()))
id_mapping = {old_id: new_id for new_id, old_id in enumerate(old_ids)}

# Obtener nombres de las clases originales
with open(f"{data_path}/dataset.yaml", "r") as f:
    dataset_yaml = yaml.safe_load(f)
class_names = dataset_yaml["names"]
new_class_names = [class_names[old_id] for old_id in old_ids]

# 2. Agrupar archivos por su clase más rara
grouped_files = {c: [] for c in sorted_classes}
for file_path, classes in file_data.items():
    # Encontrar la clase más rara presente en este archivo
    for c in sorted_classes:
        if c in classes:
            grouped_files[c].append(file_path)
            break

# 3. Partición Estratificada 90/10
train_files = []
val_files = []
for c in sorted_classes:
    files = grouped_files[c]
    random.shuffle(files)
    split_idx = max(1, int(len(files) * 0.9)) # Al menos 1 a train
    
    # Excepción si solo hay 1 archivo y queremos que también haya algo en val:
    # Como la proporción es 90/10, los grupos muy pequeños (<10) no mandarán nada a val.
    # Por seguridad con clases raras, si len(files) está entre 2 y 9, 
    # enviamos al menos 1 a val para que exista en el testeo.
    if len(files) > 1 and len(files) < 10:
        split_idx = len(files) - 1
        
    train_files.extend(files[:split_idx])
    val_files.extend(files[split_idx:])

print(f"Total a 'train': {len(train_files)} imágenes")
print(f"Total a 'val': {len(val_files)} imágenes")

# 4. Crear estructura y mover archivos
if os.path.exists(yolo_dataset_dir):
    shutil.rmtree(yolo_dataset_dir)
    
for s in ["train", "val"]:
    os.makedirs(os.path.join(yolo_dataset_dir, "images", s), exist_ok=True)
    os.makedirs(os.path.join(yolo_dataset_dir, "labels", s), exist_ok=True)

def copy_and_remap(file_list, target_split):
    copied = 0
    for file_path in file_list:
        split_orig, label_file = os.path.split(file_path)
        image_file = label_file.replace(".txt", ".jpg")
        
        src_img = os.path.join(images_dir, split_orig, image_file)
        dst_img = os.path.join(yolo_dataset_dir, "images", target_split, image_file)
        
        if os.path.exists(src_img):
            shutil.copy2(src_img, dst_img)
            
            src_lbl = os.path.join(clean_labels_dir, file_path)
            dst_lbl = os.path.join(yolo_dataset_dir, "labels", target_split, label_file)
            
            with open(src_lbl, "r") as f_in, open(dst_lbl, "w") as f_out:
                for line in f_in:
                    parts = line.strip().split()
                    if parts:
                        old_id = int(parts[0])
                        if old_id in id_mapping:
                            parts[0] = str(id_mapping[old_id])
                            f_out.write(" ".join(parts) + "\n")
            copied += 1
    return copied

print("Paso 2: Copiando y remapeando a 'train'...")
t_count = copy_and_remap(train_files, "train")
print("Paso 3: Copiando y remapeando a 'val'...")
v_count = copy_and_remap(val_files, "val")

# 5. Generar data.yaml
yaml_content = {
    "path": os.path.abspath(yolo_dataset_dir),
    "train": "images/train",
    "val": "images/val",
    "names": {i: name for i, name in enumerate(new_class_names)}
}
with open(os.path.join(yolo_dataset_dir, "data.yaml"), "w") as f:
    yaml.dump(yaml_content, f, sort_keys=False)
    
print("\n✅ Generación del Dataset Estratificado finalizada!")

# 6. Estadísticas Finales por Carpeta
print("\n📊 ESTADÍSTICAS POR CARPETA 📊")
train_stats = Counter()
val_stats = Counter()

def count_stats(split_name, stats_counter):
    label_dir = os.path.join(yolo_dataset_dir, "labels", split_name)
    if not os.path.exists(label_dir): return
    for lbl in os.listdir(label_dir):
        if not lbl.endswith(".txt"): continue
        with open(os.path.join(label_dir, lbl), "r") as f:
            for line in f:
                parts = line.strip().split()
                if parts: stats_counter[int(parts[0])] += 1

count_stats("train", train_stats)
count_stats("val", val_stats)

stats_data = []
for new_id, class_name in enumerate(new_class_names):
    train_c = train_stats[new_id]
    val_c = val_stats[new_id]
    total_c = train_c + val_c
    val_percent = (val_c / total_c * 100) if total_c > 0 else 0
    stats_data.append({
        "ID YOLO": new_id,
        "Clase": class_name,
        "Train": train_c,
        "Val": val_c,
        "Total": total_c,
        "% Val": f"{val_percent:.1f}%"
    })

df_stats = pd.DataFrame(stats_data)
# display(df_stats)
```

<p style="margin: 0 0 14px 0; line-height: 1.6; color: #cbd5e1; font-size: 0.95rem;">
Guardar cientos de miles de imágenes transformadas en disco (<em>Offline Augmentation</em>) desperdicia almacenamiento y provoca memorización estática. En pipelines de alto rendimiento aplicamos <strong>Online Data Augmentation al vuelo</strong>: las transformaciones ocurren en la memoria RAM/VRAM del <code>DataLoader</code> justo antes de alimentar la GPU, garantizando que <strong>la red nunca vea exactamente la misma imagen dos veces</strong>.

<div style="color: #94a3b8; font-size: 0.82rem; margin-top: 4px; line-height: 1.5;">Combina 4 imágenes en un mosaico unificado. Escala los objetos y los sitúa en contextos inusuales, elevando exponencialmente la detección de objetos pequeños.</div>
</div>
<div style="background: rgba(15, 23, 42, 0.6); border: 1px solid rgba(45, 212, 191, 0.2); border-radius: 8px; padding: 12px 14px;">
<strong style="color: #5eead4; font-size: 0.88rem;">Random Erasing / Cutout</strong>
<div style="color: #94a3b8; font-size: 0.82rem; margin-top: 4px; line-height: 1.5;">Enmascara parches de píxeles aleatorios, obligando a la red a identificar objetos por su estructura global en lugar de depender de una sola textura aislada.</div>
